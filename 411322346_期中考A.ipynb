````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
DB_PATH = "activity.db"   # SQLite 資料庫檔名

# 初始活動資料：(活動名稱, 日期, 名額)
INITIAL_ACTIVITIES = [
    ("迎新晚會", "2026-11-20", 6),
    ("程式工作坊", "2026-11-25", 4),
    ("籃球友誼賽", "2026-11-28", 3),
]

def init_db():
    conn = sqlite3.connect(DB_PATH)
    try:
        cur = conn.cursor()
        # 先刪報名紀錄表（它參照活動表），再刪活動表
        cur.execute("DROP TABLE IF EXISTS registrations")
        cur.execute("DROP TABLE IF EXISTS activities")

        # 活動表
        cur.execute("""
            CREATE TABLE activities (
                id       INTEGER PRIMARY KEY,
                name     TEXT UNIQUE NOT NULL,
                date     TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)

        # 報名紀錄表：people 是這筆報名的總人數（含攜伴），status 是「已報名」或「已取消」
        cur.execute("""
            CREATE TABLE registrations (
                id          INTEGER PRIMARY KEY AUTOINCREMENT,
                student     TEXT NOT NULL,
                activity_id INTEGER NOT NULL,
                people      INTEGER NOT NULL,
                status      TEXT NOT NULL,
                created_at  TEXT,
                FOREIGN KEY (activity_id) REFERENCES activities(id)
            )
        """)

        # 用 ? 參數放入初始資料
        cur.executemany(
            "INSERT INTO activities (name, date, capacity) VALUES (?, ?, ?)",
            INITIAL_ACTIVITIES
        )
        conn.commit()
    finally:
        conn.close()   # 不管有沒有出錯都關閉連線

init_db()
print("✅ 資料庫已重建，並放入 3 個活動的初始資料。")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
def show_tables():
    conn = sqlite3.connect(DB_PATH)
    try:
        print("📄 活動表 activities")
        display(pd.read_sql_query("SELECT * FROM activities", conn))

        print("📄 報名紀錄表 registrations")
        display(pd.read_sql_query("SELECT * FROM registrations", conn))
    finally:
        conn.close()   # 不管有沒有出錯都關閉連線

# 呼叫一次：此時活動表應有 3 筆，報名紀錄表應是空的
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
ACTIONS = ["報名", "取消", "查詢", "統計", "未知"]          # 允許的動作
ACTIVITIES = [a[0] for a in INITIAL_ACTIVITIES]            # 允許的活動：迎新晚會、程式工作坊、籃球友誼賽

SYSTEM_PROMPT = """你是系學會活動報名系統的解讀器。請從使用者的一句話抽出三個欄位，只輸出一行：動作,活動,人數

規則：
1. 動作只能是：報名、取消、查詢、統計、未知
   - 報名：想參加、要報名、要去
   - 取消：取消、不去了、退出
   - 查詢：問某一個活動還有幾個名額或位子
   - 統計：問所有活動或各活動的報名狀況、報名人數統計
   - 未知：和活動報名無關，或提到的活動不在下面清單中
2. 活動只能是：迎新晚會、程式工作坊、籃球友誼賽。
   可依意思對應，例如「迎新」→迎新晚會、「工作坊」→程式工作坊、「打籃球」「籃球比賽」→籃球友誼賽。
   動作是統計或未知時，活動填「無」。
3. 人數只有「報名」才需要計算，是總人數（本人＋同行的人），用阿拉伯數字。
   例如「我和兩個朋友」=3、「帶一個朋友」=2、「我們五個人」=5。沒提到人數就是 1。超過 3 也要照實寫。
   動作是取消、查詢、統計、未知時，人數填 0。
4. 只輸出一行，不要有任何說明、空格或其他文字。

範例：
使用者：我想參加迎新 → 報名,迎新晚會,1
使用者：工作坊我要參加，會帶三個同學 → 報名,程式工作坊,4
使用者：籃球比賽我要退出 → 取消,籃球友誼賽,0
使用者：程式工作坊剩多少名額 → 查詢,程式工作坊,0
使用者：幫我列出所有活動的報名人數 → 統計,無,0
使用者：我要買便當 → 未知,無,0
"""

ai_cache = {}   # 暫存：同一句話只呼叫 AI 一次，避免浪費額度


def call_ai(user_text):
    """呼叫 Groq。遇到 429（超過使用量）就等 30 秒重試，最多重試 3 次。失敗回傳 None。"""
    max_retry = 3
    for attempt in range(max_retry + 1):          # 第 1 次 + 最多重試 3 次
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": user_text},
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
            )
            time.sleep(2)                          # 每次呼叫後休息 2 秒
            return resp.choices[0].message.content or ""
        except Exception as e:
            time.sleep(2)
            is_429 = getattr(e, "status_code", None) == 429 or "429" in str(e)
            if is_429 and attempt < max_retry:
                print(f"⏳ 超過使用量（429），等待 30 秒後重試（第 {attempt + 1} 次）...")
                time.sleep(30)
            else:
                # 只印錯誤類型，避免把任何敏感資訊印出來
                print("⚠️ AI 呼叫失敗：", type(e).__name__)
                return None
    return None


def parse_reply(raw):
    """把 AI 的原始回覆轉成 (動作, 活動, 人數)，只接受允許清單裡的值。"""
    lines = [l.strip() for l in (raw or "").splitlines() if l.strip()]
    line = lines[-1] if lines else ""                 # 取最後一行（通常就是答案）
    parts = re.split(r"[,，]", line)                  # 半形、全形逗號都可以切

    action_part   = parts[0] if len(parts) >= 1 else ""
    activity_part = parts[1] if len(parts) >= 2 else line
    count_part    = parts[2] if len(parts) >= 3 else ""

    # 動作：在允許清單中找，找不到就是「未知」
    action = "未知"
    for a in ACTIONS:
        if a in action_part:
            action = a
            break

    # 活動：在允許清單中找，找不到就是 None
    activity = None
    for name in ACTIVITIES:
        if name in activity_part:
            activity = name
            break

    # 人數：用正規表示式找數字，找不到就當作 1
    m = re.search(r"\d+", count_part)
    count = int(m.group()) if m else 1

    # 需要活動的動作卻沒有合法活動 → 當作未知
    if action in ("報名", "取消", "查詢") and activity is None:
        action = "未知"
    if action in ("統計", "未知"):
        activity = "無"
    if action != "報名":
        count = 0          # 只有報名需要人數

    return action, activity, count


def ai_interpret(user_text):
    """回傳 (動作, 活動, 人數, 原始回覆)。AI 呼叫失敗時動作為「失敗」。"""
    if user_text in ai_cache:
        raw = ai_cache[user_text]
    else:
        raw = call_ai(user_text)
        if raw is None:
            return "失敗", "無", 0, None
        ai_cache[user_text] = raw                     # 成功才存進暫存
    action, activity, count = parse_reply(raw)
    return action, activity, count, raw


print("✅ AI 解讀函式已定義完成。")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
ai_tests = [
    ("我要報名迎新晚會",             "報名", "迎新晚會",   1),   # 1 正常
    ("幫我取消程式工作坊",           "取消", "程式工作坊", 0),   # 2 正常
    ("籃球友誼賽還有幾個位子？",     "查詢", "籃球友誼賽", 0),   # 3 正常
    ("我和兩個朋友要參加程式工作坊", "報名", "程式工作坊", 3),   # 4 正常（攜伴）
    ("各活動報名狀況",               "統計", "無",         0),   # 5 正常（統計）
    ("我想帶一個朋友去打籃球",       "報名", "籃球友誼賽", 2),   # 6 邊界（口語活動名＋攜伴）
    ("我們五個人要去迎新",           "報名", "迎新晚會",   5),   # 7 邊界（人數超過，交給程式拒絕）
    ("我不想去迎新晚會了",           "取消", "迎新晚會",   0),   # 8 邊界（沒說「取消」二字）
    ("今天天氣如何？",               "未知", "無",         0),   # 9 錯誤（無關的話）
    ("我要報名演唱會",               "未知", "無",         0),   # 10 錯誤（不存在的活動）
]


def same(expected, got):
    """預期值是 "*" 就算對；否則要完全相同。"""
    return expected == "*" or str(expected) == str(got)


passed = 0
for i, (text, exp_act, exp_target, exp_count) in enumerate(ai_tests, start=1):
    action, activity, count, raw = ai_interpret(text)
    ok = same(exp_act, action) and same(exp_target, activity) and same(exp_count, count)
    if ok:
        passed += 1
    print(f"{i:2d}. {'✅' if ok else '❌'} 「{text}」")
    print(f"    AI 原始回覆：{raw!r}")
    print(f"    系統解讀：{action},{activity},{count}　｜　預期：{exp_act},{exp_target},{exp_count}")

print(f"\n🎯 AI 解讀測試通過 {passed} / {len(ai_tests)} 句")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
STATUS_ON = "已報名"
STATUS_OFF = "已取消"


def get_activity(cur, activity_name):
    """找活動，回傳 (id, date, capacity)，找不到回傳 None。"""
    cur.execute("SELECT id, date, capacity FROM activities WHERE name = ?", (activity_name,))
    return cur.fetchone()


def used_seats(cur, activity_id):
    """計算某活動已被報名的人數（只算「已報名」，以人數加總）。"""
    cur.execute(
        "SELECT COALESCE(SUM(people), 0) FROM registrations WHERE activity_id = ? AND status = ?",
        (activity_id, STATUS_ON)
    )
    row = cur.fetchone()
    return row[0] if row is not None else 0


def register(student, activity_name, count):
    """報名：檢查順序 姓名 → 活動 → 人數 → 重複 → 每人上限 → 名額。"""
    if not student:
        return "❌ 報名需要姓名，請輸入姓名。"
    conn = sqlite3.connect(DB_PATH)
    try:
        cur = conn.cursor()
        act = get_activity(cur, activity_name)
        if act is None:
            return f"❌ 找不到活動「{activity_name}」。"
        activity_id, date, capacity = act

        # A1：人數必須在 1～3
        if count < 1 or count > 3:
            return f"❌ 每次報名人數須為 1～3 人（你要報名 {count} 人）。"

        # 不可重複報名同一活動（只看「已報名」，取消過的可以再報）
        cur.execute(
            "SELECT id FROM registrations WHERE student = ? AND activity_id = ? AND status = ?",
            (student, activity_id, STATUS_ON)
        )
        if cur.fetchone() is not None:
            return f"❌ {student} 已報名過「{activity_name}」，不可重複報名。"

        # A2：每人最多同時報名 2 個活動（已取消不算）
        cur.execute(
            "SELECT COUNT(*) FROM registrations WHERE student = ? AND status = ?",
            (student, STATUS_ON)
        )
        row = cur.fetchone()
        my_count = row[0] if row is not None else 0
        if my_count >= 2:
            return f"❌ 每人最多同時報名 2 個活動，{student} 目前已報名 {my_count} 個。"

        # 名額檢查（以人數加總計算）
        remaining = capacity - used_seats(cur, activity_id)
        if remaining < count:
            return f"❌ 「{activity_name}」名額不足，目前剩 {remaining} 名，你要報名 {count} 人。"

        cur.execute(
            "INSERT INTO registrations (student, activity_id, people, status, created_at) VALUES (?, ?, ?, ?, ?)",
            (student, activity_id, count, STATUS_ON, time.strftime("%Y-%m-%d %H:%M:%S"))
        )
        conn.commit()
        return f"✅ {student} 報名「{activity_name}」成功（{count} 人），目前剩餘名額 {remaining - count} 名。"
    finally:
        conn.close()


def cancel(student, activity_name):
    """取消：把紀錄改成「已取消」，名額自動釋出（因為只加總「已報名」）。"""
    if not student:
        return "❌ 取消需要姓名，請輸入姓名。"
    conn = sqlite3.connect(DB_PATH)
    try:
        cur = conn.cursor()
        act = get_activity(cur, activity_name)
        if act is None:
            return f"❌ 找不到活動「{activity_name}」。"
        activity_id, date, capacity = act

        cur.execute(
            "SELECT id, people FROM registrations WHERE student = ? AND activity_id = ? AND status = ?",
            (student, activity_id, STATUS_ON)
        )
        row = cur.fetchone()
        if row is None:
            return f"❌ 查無 {student} 在「{activity_name}」的報名紀錄。"
        reg_id, people = row

        cur.execute("UPDATE registrations SET status = ? WHERE id = ?", (STATUS_OFF, reg_id))
        conn.commit()
        remaining = capacity - used_seats(cur, activity_id)
        return f"✅ {student} 已取消「{activity_name}」，釋出 {people} 個名額，目前剩餘名額 {remaining} 名。"
    finally:
        conn.close()


def query(activity_name):
    """查詢某活動剩餘名額（不需要姓名）。"""
    conn = sqlite3.connect(DB_PATH)
    try:
        cur = conn.cursor()
        act = get_activity(cur, activity_name)
        if act is None:
            return f"❌ 找不到活動「{activity_name}」。"
        activity_id, date, capacity = act
        remaining = capacity - used_seats(cur, activity_id)
        return f"🔎 「{activity_name}」（{date}）剩餘名額 {remaining} 名（共 {capacity} 名）。"
    finally:
        conn.close()


def stats():
    """A3：用 LEFT JOIN + GROUP BY 列出每個活動的統計；沒人報名的活動也會列出（人數為 0）。"""
    conn = sqlite3.connect(DB_PATH)
    try:
        # 注意：status 條件放在 ON 裡，放在 WHERE 會把沒人報名的活動濾掉
        sql = """
            SELECT a.name, a.date, a.capacity,
                   COALESCE(SUM(r.people), 0)              AS registered,
                   a.capacity - COALESCE(SUM(r.people), 0) AS remaining
            FROM activities a
            LEFT JOIN registrations r
                   ON a.id = r.activity_id AND r.status = ?
            GROUP BY a.id, a.name, a.date, a.capacity
            ORDER BY a.id
        """
        cur = conn.cursor()
        cur.execute(sql, (STATUS_ON,))
        rows = cur.fetchall()
        lines = ["📊 各活動報名狀況"]
        for name, date, capacity, registered, remaining in rows:
            lines.append(f"・{name}｜{date}｜名額 {capacity}｜已報名 {registered}｜剩餘 {remaining}")
        return "\n".join(lines)
    finally:
        conn.close()


def list_activities():
    """無法判斷時，列出目前有哪些活動。"""
    conn = sqlite3.connect(DB_PATH)
    try:
        cur = conn.cursor()
        cur.execute("SELECT name, date FROM activities ORDER BY id")
        rows = cur.fetchall()
        lines = ["🤔 抱歉，我不太懂你的意思。目前的活動有："]
        for name, date in rows:
            lines.append(f"・{name}（{date}）")
        lines.append("你可以說：「我要報名迎新晚會」、「取消籃球友誼賽」、「程式工作坊還有幾個名額」、「各活動報名狀況」。")
        return "\n".join(lines)
    finally:
        conn.close()


print("✅ 功能函式已定義完成。")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
def handle(user_text, student=""):
    user_text = (user_text or "").strip()
    student = (student or "").strip()
    if not user_text:
        return "請輸入你想做的事，例如：「我要報名迎新晚會」。"

    action, activity, count, raw = ai_interpret(user_text)

    if action == "失敗":
        return "⚠️ 系統暫時無法處理，請稍後再試。"
    if action in ("報名", "取消") and not student:
        return "❌ 報名與取消需要姓名，請輸入姓名。"

    if action == "報名":
        return register(student, activity, count)
    if action == "取消":
        return cancel(student, activity)
    if action == "查詢":
        return query(activity)
    if action == "統計":
        return stats()
    return list_activities()        # 未知：列出目前有哪些活動


# ---- 流程測試：先重建資料庫，確保從初始狀態開始 ----
init_db()

# (姓名, 使用者說的話, 預期結果說明, 回應中應出現的關鍵字)
flow_tests = [
    ("小明", "我要報名迎新晚會",             "成功，迎新晚會剩 5 名",      "剩餘名額 5"),
    ("小明", "我要報名迎新晚會",             "拒絕：不可重複報名",          "重複"),
    ("小明", "我要報名程式工作坊",           "成功，程式工作坊剩 3 名",    "剩餘名額 3"),
    ("小明", "我也要參加籃球友誼賽",         "拒絕：每人最多 2 個活動",    "最多同時報名 2"),
    ("小華", "我和兩個朋友要參加籃球友誼賽", "成功 3 人，籃球剩 0 名",     "剩餘名額 0"),
    ("小美", "我要報名籃球友誼賽",           "拒絕：名額不足",              "名額不足"),
    ("小美", "我們五個人要去迎新晚會",       "拒絕：人數須 1～3",           "1～3"),
    ("小華", "取消籃球友誼賽",               "成功，釋出 3 個名額",         "釋出 3"),
    ("",     "籃球友誼賽還剩幾個名額？",     "籃球剩 3 名（查詢不需姓名）", "剩餘名額 3"),
    ("",     "各活動報名狀況",               "列出三個活動的統計",           "各活動報名狀況"),
]

flow_passed = 0
for i, (student, text, expected, keyword) in enumerate(flow_tests, start=1):
    result = handle(text, student)
    ok = keyword in result
    if ok:
        flow_passed += 1
    print(f"步驟 {i:2d} {'✅' if ok else '❌'}｜{student or '（空白）'}：「{text}」")
    print(f"    預期：{expected}")
    print(f"    回應：{result}\n")

print(f"🎯 流程測試通過 {flow_passed} / {len(flow_tests)} 步\n")

# 最後顯示資料表：registrations 應有 3 筆（小明×2 已報名、小華 已取消）
show_tables()
````
